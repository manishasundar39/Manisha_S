```python
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9 / 5) + 32

print(f"Temperature in Fahrenheit: {fahrenheit:.2f}")
```

```python
basic = float(input("Enter basic salary: "))

hra = basic * 0.20
da = basic * 0.15
pf = basic * 0.08
net_salary = basic + hra + da - pf

print(f"HRA: {hra:.2f}")
print(f"DA: {da:.2f}")
print(f"PF: {pf:.2f}")
print(f"Net take-home salary: {net_salary:.2f}")
```

```python
year = int(input("Enter a calendar year: "))

if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
mark1 = float(input("Enter marks for subject 1: "))
mark2 = float(input("Enter marks for subject 2: "))
mark3 = float(input("Enter marks for subject 3: "))

average = (mark1 + mark2 + mark3) / 3

if average >= 90:
    grade = "A+"
elif average >= 75:
    grade = "A"
elif average >= 50:
    grade = "B"
else:
    grade = "Fail"

print(f"Average percentage: {average:.2f}%")
print(f"Grade: {grade}")
```

```python
age = int(input("Enter your age: "))

voting_status = "Eligible to vote" if age >= 18 else "Not eligible to vote"
discount_status = "Senior citizen discount applies" if age >= 60 else "No senior citizen discount"

print(voting_status)
print(discount_status)
```

```python
number = int(input("Enter an integer: "))

for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

```python
n = int(input("Enter N: "))

total = 0
i = 1

while i <= n:
    total += i
    i += 1

if n > 0:
    average = total / n
    print(f"Sum: {total}")
    print(f"Average: {average:.2f}")
else:
    print("N must be greater than 0.")
```

```python
number = int(input("Enter an integer: "))

if number < 2:
    print("Not a Prime Number")
else:
    for i in range(2, number):
        if number % i == 0:
            print("Not a Prime Number")
            break
    else:
        print("Prime Number")
```

```python
number = int(input("Enter an integer: "))

original = abs(number)
remaining = original
reversed_number = 0

while remaining > 0:
    digit = remaining % 10
    reversed_number = reversed_number * 10 + digit
    remaining //= 10

print(f"Reversed digits: {reversed_number}")

if original == reversed_number:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
status_code = int(input("Enter an HTTP status code: "))

match status_code:
    case 200:
        print("200: OK")
    case 400:
        print("400: Bad Request")
    case 404:
        print("404: Not Found")
    case 500:
        print("500: Internal Server Error")
    case _:
        print("Unknown status code")
```
